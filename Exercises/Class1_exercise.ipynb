```python
#data types
name = "Priya"
age = 20
cgpa = 8.5
is_student = True

print(name)
print(age)
print(cgpa)
print(is_student)

print(type(name))
print(type(age))
print(type(cgpa))
print(type(is_student))
```

```python
#len, upper, title, strip
city = "chennai"
messy = "  hello  "

print(len(city))        
print(city.upper())     
print(city.title())     
print(messy.strip())    
```

```python
name = "Arun"
age = 25
print(f"Hello {name.lower()}, I'm {age} years old")
```

```python
#Multi-line f-string
name = "Arun"
age = 25
city = "Chennai"
print(f"""Hello {name}
I'm {age} years old
I live in {city}""")
```

```python
#new line and space 
print("Hello\nWorld")
print("Hello\tWorld")
```

```python
#typecasting 
mark1 = "70"
mark2 = "80"
print(mark1 + mark2)            
print(int(mark1) + int(mark2))  
```

```python
mark1 = "70.10"
mark2 = "80.10"
print(mark1 + mark2)                          
print(float(mark1) + float(mark2))            
print(int(float(mark1)) + int(float(mark2)))  
```

```python
#number to str 
a = 10.6
b = str(a)
print(b)
print(type(b), type(a))
```
