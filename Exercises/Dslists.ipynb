```python
#creating list 
list1 = [10, "Priya", 7.9, True]
print(list1)
print(type(list1))
```

```python

#first and last
nums = [10, 20, 30, 40, 50, 60]
print(nums[0])     # 10
print(nums[-1])    # 60
```

```python
#slices
nums = [10, 20, 30, 40, 50, 60]
print(nums[1:4])     # [20, 30, 40]
print(nums[::2])     # [10, 30, 50]
print(nums[::-1])    # [60, 50, 40, 30, 20, 10]
print(nums[-3:])     # [40, 50, 60]
```

```python
#change an element 
colors = ["red", "green", "blue"]
colors[1] = "yellow"
print(colors)    # ['red', 'yellow', 'blue']
```

```python
#add elements
fruits = ["apple", "banana", "cherry"]
fruits.insert(1, "mango")
print(fruits)    # ['apple', 'mango', 'banana', 'cherry']
fruits.append("grapes")
print(fruits)    # ['apple', 'mango', 'banana', 'cherry', 'grapes']
fruits.extend(["kiwi", "orange"])
print(fruits)    # ['apple', 'mango', 'banana', 'cherry', 'grapes', 'kiwi', 'orange']
```

```python
#pop and remove
nums = [10, 20, 30, 40, 50]
last = nums.pop()
print(last)      # 50
print(nums)      # [10, 20, 30, 40]
nums.remove(20)
print(nums)      # [10, 30, 40]
```

```python
#sort and sorted
marks = [75, 40, 92, 60]
print(sorted(marks))                 # [40, 60, 75, 92]
print(sorted(marks, reverse=True))   # [92, 75, 60, 40]
print(marks)                         # [75, 40, 92, 60] (unchanged)
marks.sort()
print(marks)                         # [40, 60, 75, 92] (changed)
```

```python
marks = [75, 40, 92, 60]
result = marks.sort()
print(result)    # None
print(marks)     # [40, 60, 75, 92]
```

```python
#min max len
nums = [12, 45, 7, 30]
print(min(nums))                # 7
print(max(nums))                # 45
print(len(nums))                # 4
print(sum(nums))                # 94
print(sum(nums) / len(nums))    # 23.5
```

```python
#falsey values all any 
list1 = [0, None, False, "", 10]
print(all(list1))    # False (not every value is truthy)
print(any(list1))    # True (10 is truthy)
```

```python
list1 = [10, 20, 30, []]
list2 = [[10], [None]]
print(any(list1))    # True
print(all(list1))    # False (the empty list [] is falsy)
print(any(list2))    # True
print(all(list2))    # True
```
