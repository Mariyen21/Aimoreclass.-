```python
#even or odd (if and else)
num = 17
print(num % 2 == 0)

if num % 2 == 0:
    print("number is even")
else:
    print("number is odd")
```

```python
#pass or fail 
marks = 35

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

```python
#grade (elif)
marks = 82

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 50:
    print("C")
else:
    print("Fail")
```

```python
#and,or,not 
username = "admin"
password = "abcd"

if username == "admin" and password == "1234":
    print("Login successful")
else:
    print("Invalid credentials")
```

```python
#nested if 
age = 20
has_test_passed = True

if age >= 18:
    if has_test_passed:
        print("Licence issued")
    else:
        print("Pass the test first")
else:
    print("Too young")
```

```python
# ATm Withdrawal
balance = 5000
withdraw = 4000

# remainin balance and withdrawl is successful
# insufficient

if withdraw >0:
    if withdraw <= balance:
        print("Remaining balance:",balance-withdraw)
        print("Withdraw successful")
    else:
        print("insufficient balance")
```
